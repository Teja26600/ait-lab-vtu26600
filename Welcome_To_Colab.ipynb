```
import pandas as pd
import numpy as np
from datetime import datetime, timedelta

source_cities = ['New York', 'London', 'Paris', 'Tokyo', 'Dubai', 'Singapore', 'Sydney', 'Mumbai']
destination_cities = ['London', 'Paris', 'Tokyo', 'Dubai', 'New York', 'Singapore', 'Sydney', 'Mumbai']
flight_classes = ['Economy', 'Business', 'First Class']

num_records = 50

booking_ids = [f'BK{i:05d}' for i in range(1, num_records + 1)]

sources = np.random.choice(source_cities, num_records)
destinations = np.random.choice(destination_cities, num_records)

for i in range(num_records):
    while sources[i] == destinations[i]:
        destinations[i] = np.random.choice(destination_cities)

flight_ids = [''.join(np.random.choice(list('ABCDEFGHIJKLMNOPQRSTUVWXYZ0123456789'), 6)) for _ in range(num_records)]

classes = np.random.choice(flight_classes, num_records)
amounts = np.random.randint(10000, 60001, num_records)
current_date = datetime.now()
booking_dates = [current_date - timedelta(days=np.random.randint(1, 31)) for _ in range(num_records)] # Last 30 days
travelled_dates = []

for i in range(num_records):
    travel_delay_days = np.random.randint(1, 91)
    travelled_dates.append(booking_dates[i] + timedelta(days=travel_delay_days))

data = {
    'Booking_ID': booking_ids,
    'Source': sources,
    'Destination': destinations,
    'Flight_ID': flight_ids,
    'Class': classes,
    'Amount': amounts,
    'Booking_Date': booking_dates,
    'Travelled_Date': travelled_dates
}
df_bookings = pd.DataFrame(data)

df_bookings['Booking_Date'] = df_bookings['Booking_Date'].dt.strftime('%Y-%m-%d')
df_bookings['Travelled_Date'] = df_bookings['Travelled_Date'].dt.strftime('%Y-%m-%d')

csv_file_path = 'airline_booking_records.csv'
df_bookings.to_csv(csv_file_path, index=False)

print(f"Successfully created '{csv_file_path}' with {num_records} booking records.")
print("First 5 rows of the generated DataFrame:")
print(df_bookings.head())
```

```
import numpy as np

try:
    amount_data = np.loadtxt(
        'airline_booking_records.csv',
        delimiter=',',
        skiprows=1,
        usecols=(5,)
    )
    print("Successfully loaded 'Amount' column using numpy.loadtxt().")
    print("First 5 elements of 'Amount' data:")
    print(amount_data[:5])
except Exception as e:
    print(f"Error loading with numpy.loadtxt() due to mixed data types (as expected):\n{e}")
```

```
print("\nFirst 5 rows of the data loaded with numpy.genfromtxt():")
try:
    data_genfromtxt = np.genfromtxt(
        'airline_booking_records.csv',
        delimiter=',',
        skip_header=1,
        dtype=None, # Allow for mixed data types
        encoding='utf-8'
    )
    print("Successfully loaded data using numpy.genfromtxt().")
    # genfromtxt returns a 1D array of tuples if dtype=None and records are not uniform
    # If it's a 2D array, slice directly. If 1D array of tuples, convert to list for printing first few.
    if data_genfromtxt.ndim == 1 and data_genfromtxt.dtype.names is not None:
        # It's a structured array, print first 5 records
        for i, row in enumerate(data_genfromtxt[:5]):
            print(row)
    elif data_genfromtxt.ndim > 0:
        # It's a regular array, print first 5 rows
        print(data_genfromtxt[:5])
    else:
        print(data_genfromtxt)
except Exception as e:
    print(f"Error loading with numpy.genfromtxt():\n{e}")
```
